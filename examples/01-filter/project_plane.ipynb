Project to a Plane {#project_plane_example}
==================

Project a `~pyvista.PolyData`{.interpreted-text role="class"} surface or
pointset to a plane.

Uses
`~pyvista.PolyDataFilters.project_points_to_plane`{.interpreted-text
role="meth"} with a normal and origin.


```python
import pyvista as pv
from pyvista import examples

poly = examples.load_random_hills()
poly.plot()
```

Project that surface to a plane underneath the surface


```python
origin = list(poly.center)
origin[-1] -= poly.length / 3.0
projected = poly.project_points_to_plane(origin=origin)

# Display the results
pl = pv.Plotter()
pl.add_mesh(poly)
pl.add_mesh(projected)
pl.show()
```
