Remove Cells {#remove_cells_example}
============

Remove specific cells from a mesh.

Cells can be removed from any dataset by index or with a boolean mask
using `~pyvista.DataSetFilters.remove_cells`{.interpreted-text
role="meth"}. The output is a `~pyvista.PolyData`{.interpreted-text
role="class"} for `PolyData` input and an
`~pyvista.UnstructuredGrid`{.interpreted-text role="class"} otherwise.


```python
from pyvista import examples

mesh = examples.load_channels()
```

Decide which cells to remove with a criterion (feel free to adjust this
or manually create this array to remove specific cells).


```python
remove = mesh['facies'] < 1.0
```

Remove the cells and plot the result.


```python
mesh = mesh.remove_cells(remove)
mesh.plot(clim=[0, 4])
```
